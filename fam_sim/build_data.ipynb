```python
from sincfold.dataset import SeqDataset, pad_batch
from torch.utils.data import DataLoader
import pandas as pd
```

```python
train_file = (
    "/home/gkulemeyer/Documents/Repos/sincFold/fam_sim/data/fam-fold/train_5s.csv"
)

data = pd.read_csv(train_file)
valid_split = 0.1
# train_file = os.path.join(out_path, "train.csv")
# valid_file = os.path.join(out_path, "valid.csv")

val_data = data.sample(frac=valid_split)
# val_data.to_csv(valid_file, index=False)
data.drop(val_data.index).to_csv(train_file, index=False)

batch_size = 4
train_loader = DataLoader(
    SeqDataset(train_file, training=True),
    batch_size=batch_size,
    shuffle=True,
    num_workers=2,
    collate_fn=pad_batch,
)
```

```python
steps = 1500
ep = int(round(steps / len(train_loader)))
ep = steps // len(train_loader)
print(len(train_loader))
print(ep)
```
