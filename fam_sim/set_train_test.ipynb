```python
import os

fams = ["5s", "tmRNA", "tRNA", "srp", "grp1", "RNaseP", "23s", "telomerase", "16s"]
DATA_PATH = "data/fam-fold/"
DATA200_PATH = "data/rnadist/dist_200"
DATA400_PATH = "data/rnadist/dist_400"

MODELS_PATH = "models/fam-sim/5_epochs/"
MODELS200_PATH = "models/rnadist/dist_200/5_epochs/"
MODELS400_PATH = "models/rnadist/dist_200/5_epochs/"

os.makedirs(MODELS_PATH, exist_ok=True)
os.makedirs(MODELS200_PATH, exist_ok=True)
os.makedirs(MODELS400_PATH, exist_ok=True)


for fam in fams:
    train_file = f"{DATA_PATH}train_{fam}.csv"
    valid_file = f"{DATA_PATH}valid_{fam}.csv"
    out_path = f"{MODELS_PATH}{fam}/"
    train = f"sincFold -d cuda train {train_file} --valid-file {valid_file} -o {out_path} -n 5"
    print(f"Running: {train}")
    os.system(train)
```
